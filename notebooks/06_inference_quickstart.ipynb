# Persian TTS Quickstart Demo
Synthesize Persian text to speech using pretrained VITS in just a few lines of code.

```python
from persian_tts import PersianTTS
import IPython.display as ipd

tts = PersianTTS()
text = 'سلام، به سامانه سنتز گفتار زبان فارسی خوش آمدید.'
audio = tts.synthesize(text)
ipd.Audio(audio, rate=16000)
```
