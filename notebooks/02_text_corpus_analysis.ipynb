```python
pd.set_option('display.max_colwidth', None)
```

```python
file= '/kaggle/input/persian-tts-dataset-famale/metadata.csv'
```

```python
unwanted_chars = {'.','،','!','?','؟','<','>','k','un',}
```

```python
metadata = pd.read_csv(file , header=None) 
```

```python
# grapheme_chars = set()
# for text in grapheme :
#     grapheme_chars.update(list(text))
# grapheme_chars = sorted(list(grapheme_chars))
# print(grapheme_chars)
```

```python
metadata[['text', 'file']] = metadata[0].str.split('|', expand=True)
metadata = metadata.drop(columns = [0])
pattern = "[" + re.escape("".join(unwanted_chars)) + "]"
metadata['text'] = metadata['text'].apply(lambda x: re.sub(pattern, ' ', x))
# metadata
```

```python
metadata['tokens'] = metadata['text'].apply(lambda words: [char for word in words for char in word])
# metadata
```

```python
metadata= metadata.drop(columns=['text'])
# metadata
```

```python
# metadata['tokens']
all_tokens = [token for row in metadata['tokens'] for token in row]
unique_tokens = set(all_tokens)
print(unique_tokens)
```
